```python
import json
import random
from sklearn.naive_bayes import MultinomialNB
from sklearn.feature_extraction.text import TfidfVectorizer
import joblib

with open("new.json","r") as file :
    data= json.load(file)
# print(data)    
#arrange data
#pattern =[hi]
#tag =[g]
#res={tag:response}

sentance=[]
lables=[]
response={}

for i in data ['intents']:
    tag=i['tag']
    response[tag]=i["responses"]
    for pattern in i['patterns']:
        sentance.append(pattern)
        lables.append(tag)
print(sentance) 
print(lables)  
print(response)  

vector=TfidfVectorizer()
x=vector.fit_transform(sentance)
model=MultinomialNB()
model.fit(x,lables)
joblib.dump(model,"chatbot.pkl")
joblib.dump(vector,"vector.pkl")
joblib.dump(response,"res.pkl")


```

```python
import joblib
import random
model=joblib.load("chatbot.pkl")
vector=joblib.load("vector.pkl")
response=joblib.load("res.pkl")
while True:
    user=input("You:")
    if user== "exit":
        print("Bot:Thank you")
        break
    user_vector=vector.transform([user]).toarray()
    predict = model.predict(user_vector)[0]
    if predict in response:
         reply=random.choice(response[predict])
         print("Bot:",reply)
```
