```python
import joblib
import random
model=joblib.load("chatbot.pkl")
vector=joblib.load("vector.pkl")
response=joblib.load("res.pkl")
while True:
    user=input("You:")
    if user== "exit":
        print("Bot:Thank you")
        break
    user_vector=vector.transform([user]).toarray()
    predict = model.predict(user_vector)[0]
    if predict in response:
         reply=random.choice(response[predict])
         print("Bot:",reply)
```
